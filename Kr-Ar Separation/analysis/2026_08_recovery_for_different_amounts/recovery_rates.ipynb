```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.axes as ax
import scipy.optimize as opt
import pandas as pd
```

```python
# data frame
df = pd.read_csv("/Users/fatmayakut/Desktop/Physik/MA/Kr-Ar Separation/analysis/2026_08_recovery_for_different_amounts/volume_v_recovery.txt", sep="\t", header=0, index_col=0)
#df = df.groupby("Input")["Recovery"].agg(["mean", "std", "count"])
x = df.index.values
y = df["Recovery"].values*100#df["mean"].values

l = len(df.index.values)
x2 = np.linspace(0, 25, l)
y2 = np.ones(l)*100
y3 = np.ones(l)*50
y4 = np.ones(l)*90
#dy = df["std"].values / np.sqrt(df["count"].values)  # standard error
```

```python
# plot
plt.figure(figsize=(8, 6))

plt.errorbar(x, y, fmt='o', capsize=3) #yerr=dy

plt.plot(x2, y2, color='r', label='100% Recovery')
plt.plot(x2, y4, color='pink', label='90% Recovery')
plt.plot(x2, y3, color='mistyrose', label='50% Recovery')
plt.fill_between(x, y2, y4, color='pink', alpha=0.2)

plt.yscale('linear')
plt.ylim(0, 110)
plt.xlim(0, 25)
plt.xlabel("Volume (uL)")
plt.ylabel("Recovery (%)")
plt.title("Recovery vs Volume")
plt.legend(loc='lower right')
plt.show()
```
