```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.axes as ax
import scipy.optimize as opt
import pandas as pd
```

```python
# data frame
df = pd.read_csv("/Users/fatmayakut/Desktop/Physik/MA/Kr:Ar Separation/analysis/2026_08_recovery_for_different_amounts/volume_v_recovery.txt", sep="\t", skiprows=[0, 2, 6], names=["Input", "Recovery"])
#df = df.groupby("Input")["Recovery"].agg(["mean", "std", "count"])
x = df.index.values
y = df["Recovery"]#df["mean"].values
#dy = df["std"].values / np.sqrt(df["count"].values)  # standard error
```

```python
# plot
plt.figure(figsize=(8, 6))

plt.errorbar(x, y, fmt='o', label='Data', capsize=5) #yerr=dy

plt.yscale('linear')
plt.ylim(0, 1)
plt.xlabel("Volume (uL)")
plt.ylabel("Recovery (%)")
plt.title("Recovery vs Volume")
plt.legend()
plt.show()
```
