```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.axes as ax
import scipy.optimize as opt
import pandas as pd
```

```python
# data frame
df = pd.read_csv("/Users/fatmayakut/Desktop/Physik/MA/Kr-Ar Separation/analysis/2026_08_recovery_for_different_amounts/method_1.txt", sep="\t", header=0, index_col=0)
#df = df.groupby("Input")["Recovery"].agg(["mean", "std", "count"])
x = df['Kr in'].values
x50 = x[df["recovery"].values < 0.9]
x90 = x[df["recovery"].values >= 0.9]
y = df["recovery"].values*100#df["mean"].values
y50 = df["recovery"].values[df["recovery"].values < 0.9]*100
y90 = df["recovery"].values[df["recovery"].values >= 0.9]*100

l = len(df.index.values)
x2 = np.linspace(0, 25, l)
y2 = np.ones(l)*100
y3 = np.ones(l)*50
y4 = np.ones(l)*90
#dy = df["std"].values / np.sqrt(df["count"].values)  # standard error
```

```python
# plot
plt.figure(figsize=(10, 6))

plt.plot(x2, y2, color='darkgreen', label='100%')

plt.errorbar(x90, y90, color='blue', fmt='o', capsize=2, label=r'$\geq$90% Recovery') #yerr=dy
plt.errorbar(x50, y50, color='salmon', fmt='o', capsize=2, markeredgecolor='firebrick', alpha=0.8, label=r'$\geq$50% Recovery') #yerr=dy

#c1 = plt.scatter(Dose_SAS, SF_SAS_ATN200_mean, color='darkorange', edgecolors = 'black', s = 70, label=f'SAS ATN 200 $\mu$M')


plt.fill_between(x2, y2, y4, color='yellowgreen', alpha=0.2)
#plt.plot(x2, y4, color='pink', label='90% Recovery', alpha=0.2)
plt.plot(x2, y3, color='firebrick')


plt.yscale('linear')
plt.ylim(40, 105)
plt.xlim(0, 25)
plt.xlabel(r"Kr Volume [$\mu$L STP]", fontsize=15)
plt.ylabel("Recovery [%]", fontsize=15)
plt.title("Recovery vs Volume", fontsize=18)
plt.legend(loc='center right')
plt.grid(alpha=0.3)
plt.savefig("/Users/fatmayakut/Desktop/Physik/MA/Kr-Ar Separation/analysis/2026_08_recovery_for_different_amounts/not_H2_corrected.png", dpi=300, bbox_inches='tight')
plt.show()
```
